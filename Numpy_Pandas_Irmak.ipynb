```python
# 🤖 NumPy ve Pandas Uygulamalı Etkinliği - Yapay Zekâya Giriş
# --------------------------------------------------------------
# Bu çalışmada NumPy ve Pandas kütüphanelerini uygulamalı olarak öğreneceksiniz.
# Kodları kendiniz yazarak parametrelerin ne işe yaradığını anlamaya çalışın.
# Her görevden sonra, kodunuzu ve açıklamanızı ilgili alana yazın.

# =============================================================
# 1. NUMPY - SAYISAL VERİ İŞLEME
# =============================================================

# 1.1 NumPy Kütüphanesini Dahil Et
# Görev: NumPy kütüphanesini içe aktarın ve sürümünü ekrana yazdırın.
# Soru: 'as np' ifadesi ne işe yarıyor olabilir?

# 💻 Kodunuzu buraya yazın:
import numpy as np #kütüphane eklio
print(np.__version__)# sürüm yazıo



# 📝 Cevap: diğer sütünlarda tam ismi kullanmak yerine np diye kısaltıp yazamızı sağlar (işi kolaylaştırır.)


# --------------------------------------------------------------
# 1.2 NumPy Dizisi Oluşturma
# Görev: 0'dan 10'a kadar sayılardan oluşan bir NumPy dizisi oluşturun.
# Soru: np.arange() içindeki parametreler neyi belirler?

# 💻 Kodunuzu buraya yazın:
import numpy as np
dizi = np.arange(0, 11)#diziyi kuruo
print(dizi)#diziyi yazdırıo


# 📝 Cevap: kaçtan kaça dizeyi kuracağını belirler ama son sayıdan -1 olur sonu çünkü 0 dan başlar 0 yerine 1 koyulursa tam 11 e kadar dizi kurar


# --------------------------------------------------------------
# 1.3 Diziyi İnceleme
# Görev: Dizinin boyutu, eleman sayısı ve veri tipini yazdırın.
# Soru: ndim, size ve dtype ne işe yarar?

# 💻 Kodunuzu buraya yazın:
import numpy as np
dizi = np.arange(0, 11)
print("Boyut:", dizi.ndim)# boyut verio 
print("Eleman sayısı:", dizi.size)# E sayısı verio
print("Veri tipi:", dizi.dtype)# veri tipi belirtiı



# 📝 Cevap: ndim 1d 2d 3d olduğunu söyler size son sayıda kaç yazıyorsa onu söyler dtype sayısalsas int  metinselse string olduğuunu belirtir


# --------------------------------------------------------------
# 1.4 Sıfır ve Bir Dizileri
# Görev: 2 satır, 4 sütundan oluşan sıfırlardan ve birlerden diziler oluşturun.
# Soru: np.zeros((satir, sutun)) yapısındaki parantez neden iki katlı?

# 💻 Kodunuzu buraya yazın:
import numpy as np
sifir_dizi = np.zeros((2, 4))#sıfır diziasi eklio
bir_dizi = np.ones((2, 4))#birler dizisini kuruo
print("Sıfır dizisi:", sifir_dizi)#sısfır dizi ekrana yansıtma 
print("Bir dizisi:", bir_dizi)#birler hali


# 📝 Cevap:çok boyutlu olduğu için - 2d yani


# --------------------------------------------------------------
# 1.5 Rastgele Sayılardan Dizi
# Görev: 0 ile 100 arasında rastgele 5 sayıdan oluşan bir dizi oluşturun.
# Soru: randint() fonksiyonundaki parametreler neyi kontrol eder?

# 💻 Kodunuzu buraya yazın:
import numpy as np
dizi = np.random.randint(0, 101, 5)#rastgele sayı üreten kod
print(dizi)#rastgele yazıları yazıo


# 📝 Cevap:başlayan sayı ile son sayı arasında kaç sayı üreteleceğini 


# --------------------------------------------------------------
# 1.6 Dizilerle İşlemler
# Görev: İki farklı NumPy dizisi (a ve b) oluşturup toplama, çıkarma, çarpma, bölme yapın.
# Soru: Bu işlemler eleman bazında mı yoksa dizi bazında mı olur?

# 💻 Kodunuzu buraya yazın:
import numpy as np
a = np.array([10, 20, 30])
b = np.array([1, 2, 3])
print("Toplama:", a + b)#toplama
print("Çıkarma:", a - b)#çıkarma
print("Çarpma:", a * b)#çarpma
print("Bölme:", a / b)#bölme


# 📝 Cevap:eleman


# --------------------------------------------------------------
# 1.7 İndeksleme ve Dilimleme
# Görev: 10'dan 20'ye kadar sayılardan oluşan bir dizi oluşturun.
#  - İlk 4 elemanı, son 2 elemanı ve ortadaki 3 elemanı seçin.
# Soru: [baslangic:bitis] yapısındaki sayılar neyi belirler?

# 💻 Kodunuzu buraya yazın:
import numpy as np
dizi = np.arange(10, 21)
print(dizi)
print(dizi[:4])#ilk dört
print(dizi[-2:])#son iki sondan almak için (-) koyulur 
print(dizi[3:6])# (:) ortaya koyunca orta verio 


# 📝 Cevap:Diziden hangi indeks aralığının seçileceğini belirlermiş


# --------------------------------------------------------------
# 1.8 Basit İstatistiksel İşlemler
# Görev: 6 nottan oluşan bir NumPy dizisi oluşturup ortalama, maksimum, minimum bulun.
# Soru: np.mean(), np.max(), np.min() fonksiyonlarının parametreleri ne işe yarar?

# 💻 Kodunuzu buraya yazın:
import numpy as np
notlar = np.array([70, 85, 90, 60, 75, 95])
ortalama = np.mean(notlar)
maksimum = np.max(notlar)
minimum = np.min(notlar)
print("Ortalama:", ortalama)
print("Maksimum:", maksimum)
print("Minimum:", minimum)



# 📝 Cevap:mean ortalama alıo max en fazlasını min en azını buluo


# =============================================================
# 2. PANDAS - TABLOLARLA VERİ ANALİZİ
# =============================================================

# 2.1 Pandas Kütüphanesini Dahil Et
# Görev: Pandas kütüphanesini içe aktarın ve sürümünü yazdırın.
# Soru: 'as pd' neden kullanılır?

# 💻 Kodunuzu buraya yazın:
import pandas as pd 
print(pd.__version__)


# 📝 Cevap: pandas uzuzun yazmak yerine kısaltıo yazıoz kolaylık sağlıo


# --------------------------------------------------------------
# 2.2 Basit Bir Tablo (DataFrame) Oluşturma
# Görev: Aşağıdaki bilgileri kullanarak bir tablo oluşturun:
# Ad | Yaş | Not
# Ali | 17 | 85
# Zeynep | 18 | 90
# Ayşe | 17 | 70
# Soru: DataFrame yapısında satır ve sütunlar nasıl gösterilir?

# 💻 Kodunuzu buraya yazın:
import pandas as pd
veri = {
    "Ad": ["Ali", "Zeynep", "Ayşe"],
    "Yaş": [17, 18, 17],
    "Not": [85, 90, 70]
}
df = pd.DataFrame(veri)
print(df)


# 📝 Cevap:print(df.columns) sütün ları verio index de satırları 


# --------------------------------------------------------------
# 2.3 Sütun ve Satır Seçme
# Görev: Sadece 'Ad' sütununu ve 2. satırı görüntüleyin.
# Soru: iloc[] ve loc[] arasındaki fark nedir?

# 💻 Kodunuzu buraya yazın:
import pandas as pd
df = pd.DataFrame({
    "Ad": ["Ali", "Zeynep", "Ayşe"],
    "Yaş": [17, 18, 17],
    "Not": [85, 90, 70]
})
print("Ad sütunu:", df["Ad"])
print("2. satır:", df.iloc[1])

# 📝 Cevap:iloc sayısal seçim loc ise isim etiket seçim yapar


# --------------------------------------------------------------
# 2.4 Filtreleme
# Görev: Notu 80'den büyük olan öğrencileri listeleyin.
# Soru: Filtreleme işlemlerinde >, <, == sembolleri nasıl kullanılır?

# 💻 Kodunuzu buraya yazın:
import pandas as pd
df = pd.DataFrame({
    "Ad": ["Ali", "Zeynep", "Ayşe"],
    "Yaş": [17, 18, 17],
    "Not": [85, 90, 70]
})
filtre = df[df["Not"] > 80]
print(filtre)


# 📝 Cevap:nottan sonra ne isteliniosa onun sembolü koyulur 


# --------------------------------------------------------------
# 2.5 Ortalama Hesaplama
# Görev: Sınıfın ortalama notunu hesaplayın.
# Soru: mean() fonksiyonu sayısal olmayan sütunlarda çalışır mı?

# 💻 Kodunuzu buraya yazın:
import pandas as pd
df = pd.DataFrame({
    "Ad": ["Ali", "Zeynep", "Ayşe"],
    "Yaş": [17, 18, 17],
    "Not": [85, 90, 70]
})
ortalama_not = df["Not"].mean()
print("Sınıfın ortalama notu:", ortalama_not)



# 📝 Cevap:hata veriyormuş yani çalışmıo


# --------------------------------------------------------------
# 2.6 Ek Görev (isteğe bağlı)
# Görev: 3 kitap ve fiyatından oluşan bir tablo oluşturun, en pahalı kitabı bulun.
# Soru: sort_values() fonksiyonundaki ascending parametresi ne işe yarar?

# 💻 Kodunuzu buraya yazın:
import pandas as pd
df = pd.DataFrame({
    "Kitap": ["Kitap A", "Kitap B", "Kitap C"],
    "Fiyat": [50, 80, 60]
})
# En pahalı kitabı bulmak için yazılan kod 
en_pahali = df.sort_values("Fiyat", ascending=False).iloc[0]
print("En pahalı kitap:", en_pahali)



# 📝 Cevap:false yazınca yüksekten küçüğe yapmaya true yapınca düşükten yükseğe yapmaya yarıo


# =============================================================
# Teslim Kuralları
# - Kodları kendiniz yazın ve açıklama satırları ekleyin.
# - Dosya adı: NumPy_Pandas_AdSoyad.ipynb
# - Google Colab üzerinde çalışın.
# =============================================================
```
